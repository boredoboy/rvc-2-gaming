# RVC Realtime — Kaggle GPU

Для простого запуска нажмите Run у ячейки в интерактивной Kaggle-сессии и оставьте вкладку открытой. Если хотите закрыть редактор, используйте Save Version → Save & Run All: он создаёт отдельную сессию. В обоих режимах включите Internet и GPU T4, затем разрешите secrets `RVC_ACCESS_PASSWORD`, `RVC_TELEGRAM_BOT_TOKEN`, `RVC_TELEGRAM_CHAT_ID`. Internet обязателен для загрузки зависимостей и работы сайта/аудио-туннеля.

При первом запуске устанавливаются Python 3.10, RVC inference dependencies и браузерный клиент. После остановки отправьте Telegram боту `/stop`, затем остановите Kaggle-сессию, чтобы освободить GPU.

```python
from pathlib import Path
import shutil
import subprocess

project = Path('/kaggle/working/rvc2-project')
if project.exists():
    shutil.rmtree(project)
subprocess.run([
    'git', 'clone', '--depth', '1',
    'https://github.com/boredoboy/rvc-2-gaming.git', str(project)
], check=True)
subprocess.run([
    'python', str(project / 'scripts' / 'run_kaggle.py')
], check=True)
```
